```python
from biolearn.data_library import DataLibrary
```

```python
GSEs = [
    "GSE109627", "GSE195834", "GSE66351", "GSE125895", "GSE134379", "GSE76105", "GSE61380", "GSE61107",
    "GSE74193", "GSE80970", "GSE144858", "GSE53740", "GSE190540", "GSE111629", "GSE122244"
]
```

```python
loaded_models = {}
for gse in GSEs:
    print(f"Loading {gse}...")
    data_source = DataLibrary().get(gse)
    loaded_models[gse] = data_source
```

```python
missing = []

for key in sorted(loaded_models.keys()):
    print(f"Checking {key}...")
    print(loaded_models[key])
    try:
        loaded_models[key].load()
    except Exception as e:
        print(f"Failed to load {key}: {e}")
        missing.append(key)
```
